```python
for col in df. columns:
    # print(f"Column: {col}, Type: {df[col].dtype}")
    print(f"{col},Type:{df[col].dtype}")
```

```python
record=[]
for col in df.columns:
    n_unique=df[col].nunique(dropna=True)
    # print(f"{col}: {n_unique}")
    sample=df[col].dropna().iloc[0] if df[col].notna().any else None
    # print(f"Sample value: {sample}")
    stored_type=str(df[col].dtype)

    is_suspicios= False
    if df[col].dtype=='object':
        converted=pd.to_numeric(df[col], errors='coerce')
        pct_covertible=converted.notna().mean()
        if pct_covertible>0.9:
            is_suspicios=True
    # print(f"Is suspicious: {is_suspicios}")
    record.append({
        "column": col,
        "stored_type":stored_type,
        "n_unique": n_unique,
        "sample_value": sample, 
        "dtype_suspicious": is_suspicios})
print(pd.DataFrame(record))    
```
